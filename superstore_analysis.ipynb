```python
df["profit_margin"]=(df["Profit"]/df["Sales"])*100
```

```python
df[["Sales", "Profit", "profit_margin"]].head()
```

```python
def discount_category(discount):
    if discount == 0:
        return "No Discount"
    elif discount <= 0.20:
        return "Low Discount"
    elif discount <= 0.40:
        return "Medium Discount"
    else:
        return "High Discount"

df["Discount_Category"] = df["Discount"].apply(discount_category)
```

```python
df["Discount_Category"].value_counts()
```

```python
df["Profit_Status"] = df["Profit"].apply(
    lambda x: "Profit" if x > 0 else "Loss"
)
```

```python
df["Profit_Status"].value_counts()
```

```python
total_sales=df["Sales"].sum()
print(total_sales)
```

```python
total_profit=df["Profit"].sum()
print(total_profit)
```

```python
total_Quantity=df["Quantity"].sum()
print(total_Quantity)
```

```python
avg_sales=df["Sales"].mean()
print(avg_sales)
```

```python
avg_profit=df["Profit"].mean()
print(avg_profit)
```

```python
profit_margin = (total_profit / total_sales) * 100
print("Profit Margin:", profit_margin)
```

```python
kpi={ "Total Sales": total_sales,
    "Total Profit": total_profit,
    "Total Quantity": total_Quantity,
    "Average Sales": avg_sales,
    "Average Profit": avg_profit,
    "Profit Margin %": profit_margin
}
```

```python
category_analysis = (
    df.groupby("Category")
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum"),
          Quantity=("Quantity", "sum")
      )
      .sort_values("Sales", ascending=False)
)

category_analysis
```

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_quantity = df["Quantity"].sum()

profit_margin = (total_profit / total_sales) * 100

loss_orders = (df["Profit"] < 0).sum()

print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
print("Total Quantity:", total_quantity)
print("Profit Margin:", profit_margin)
print("Loss-Making Orders:", loss_orders)
```

```python
print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total Quantity: {total_quantity:,}")
print(f"Profit Margin: {profit_margin:.2f}%")
print(f"Loss-Making Orders: {loss_orders:,}")
```

```python
fig, ax = plt.subplots(figsize=(12, 6))

ax.axis("off")

kpis = [
    ("Total Sales", f"${total_sales:,.0f}"),
    ("Total Profit", f"${total_profit:,.0f}"),
    ("Total Quantity", f"{total_quantity:,}"),
    ("Profit Margin", f"{profit_margin:.2f}%"),
    ("Loss Orders", f"{loss_orders:,}")
]

for i, (label, value) in enumerate(kpis):
    ax.text(
        i * 0.25 + 0.05,
        0.55,
        label,
        ha="center",
        fontsize=12
    )

    ax.text(
        i * 0.25 + 0.05,
        0.40,
        value,
        ha="center",
        fontsize=18,
        fontweight="bold"
    )

plt.title(
    "Superstore Sales & Profit Analysis — KPI Summary",
    fontsize=18,
    fontweight="bold"
)

plt.show()
```
